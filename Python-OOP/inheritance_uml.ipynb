# BÖLÜM 5 — super() neden var?

Önce doğru kod:

````python
class Kisi:
    def __init__(self, ad, soyad, email):
        self.ad = ad
        self.soyad = soyad
        self.email = email


class Ogrenci(Kisi):
    def __init__(self, ad, soyad, email, numara):

        super().__init__(ad, soyad, email)

        self.numara = numara
````

Şu satırı parçalayalım:

```python
super().__init__(ad, soyad, email)
```

### `super()`

Üst sınıf tarafındaki davranışlara erişmemizi sağlar.

### `__init__`

Üst sınıfın kurucu metodunu çağırıyoruz.

### `(ad, soyad, email)`

Üst sınıfın ihtiyaç duyduğu değerleri gönderiyoruz.

Yani öğrenci oluşturulurken kabaca şu gerçekleşir:

```text
Ogrenci.__init__()
        |
        | super().__init__()
        v
Kisi.__init__()
        |
        ├── self.ad oluştur
        ├── self.soyad oluştur
        └── self.email oluştur
        |
        v
Ogrenci.__init__() devam eder
        |
        └── self.numara oluştur
```

````python
ogrenci1 = Ogrenci(
    "Ayşe",
    "Yılmaz",
    "ayse@ogr.edu.tr",
    2026001
)

print("Ad:", ogrenci1.ad)
print("Soyad:", ogrenci1.soyad)
print("E-posta:", ogrenci1.email)
print("Numara:", ogrenci1.numara)
````

# BÖLÜM 6 — İkinci alt sınıfı ekleyelim

Şimdi `Akademik` de `Kisi` sınıfından miras alsın.

````python
class Akademik(Kisi):
    def __init__(self, ad, soyad, email, unvan):

        super().__init__(ad, soyad, email)

        self.unvan = unvan
````

````python
ogrenci = Ogrenci(
    "Ayşe",
    "Yılmaz",
    "ayse@ogr.edu.tr",
    2026001
)

akademik = Akademik(
    "Mehmet",
    "Demir",
    "mehmet@uni.edu.tr",
    "Doç. Dr."
)

print("ÖĞRENCİ")
print(ogrenci.ad)
print(ogrenci.numara)

print("\nAKADEMİK")
print(akademik.ad)
print(akademik.unvan)
````

```text

        ad, soyad, email
                  |
          +-------+-------+
          |               |
       Ogrenci         Akademik
       numara           unvan
```

Her iki alt sınıf ortak bilgileri `Kisi`den alıyor.

Kendilerine özgü bilgileri ise kendileri ekliyor.

# BÖLÜM 7 — Üst sınıfın metodunu kullanmak

`Kisi` sınıfına tekrar bir metot ekleyelim.

````python
class Kisi:
    def __init__(self, ad, soyad, email):
        self.ad = ad
        self.soyad = soyad
        self.email = email

    def tam_ad(self):
        return self.ad + " " + self.soyad

    def bilgi_goster(self):
        print("Ad Soyad:", self.tam_ad())
        print("E-posta:", self.email)


class Ogrenci(Kisi):
    def __init__(self, ad, soyad, email, numara):
        super().__init__(ad, soyad, email)
        self.numara = numara


ogrenci = Ogrenci(
    "Ayşe",
    "Yılmaz",
    "ayse@ogr.edu.tr",
    123
)

print(ogrenci.tam_ad())
ogrenci.bilgi_goster()
````

`Ogrenci` içinde `tam_ad()` yok.

`bilgi_goster()` da yok.

Ama ikisini de kullanabiliyor.

Bu yüzden kalıtım yalnızca attribute mirası değildir.

Metotlar da miras alınabilir.

# BÖLÜM 8 — Miras alınan metot her zaman yeterli mi?

`Kisi.bilgi_goster()`:

- ad soyad
- email

gösteriyor.

Ama öğrenci için ayrıca numarayı göstermek istiyoruz.

Bir seçenek yeni isimli bir metot yazmak:

```python
ogrenci_bilgisi_goster()
```

Ama daha iyi bir seçenek aynı `bilgi_goster()` metodunu öğrenciye uygun hale getirmektir.

Buna **method overriding** denir.

````python
class Kisi:
    def __init__(self, ad, email):
        self.ad = ad
        self.email = email

    def bilgi_goster(self):
        print("Ad:", self.ad)
        print("E-posta:", self.email)


class Ogrenci(Kisi):
    def __init__(self, ad, email, numara):
        super().__init__(ad, email)
        self.numara = numara

    def bilgi_goster(self):
        print("Ad:", self.ad)
        print("E-posta:", self.email)
        print("Öğrenci No:", self.numara)


ogrenci = Ogrenci(
    "Ayşe",
    "ayse@ogr.edu.tr",
    123
)

ogrenci.bilgi_goster()
````

Çalıştı ama yine tekrar oluştu:

```python
print("Ad:", self.ad)
print("E-posta:", self.email)
```

Üst sınıf zaten bunları yapıyordu.

Bir metodun içerisinden de `super()` kullanabiliriz.

````python
class Kisi:
    def __init__(self, ad, email):
        self.ad = ad
        self.email = email

    def bilgi_goster(self):
        print("Ad:", self.ad)
        print("E-posta:", self.email)


class Ogrenci(Kisi):
    def __init__(self, ad, email, numara):
        super().__init__(ad, email)
        self.numara = numara

    def bilgi_goster(self):

        super().bilgi_goster()

        print("Öğrenci No:", self.numara)


ogrenci = Ogrenci(
    "Ayşe",
    "ayse@ogr.edu.tr",
    123
)

ogrenci.bilgi_goster()
````

Burada iki farklı `super()` kullanımı gördük:

```python
super().__init__(...)
```

→ üst sınıfın kurucusunu kullan.

```python
super().bilgi_goster()
```

→ üst sınıfın `bilgi_goster()` metodunu kullan.

Sonra alt sınıfa özgü davranışı ekle.

# BÖLÜM 9 — Çok basit ikinci örnek: Hayvanlar

Şimdi aynı kavramı farklı bir problemde görelim.

Ortak özellik:

- ad

Ortak davranış:

- yemek_ye()

Özel davranışlar:

- Kedi → miyavla()
- Kopek → havla()

````python
class Hayvan:
    def __init__(self, ad):
        self.ad = ad

    def yemek_ye(self):
        print(self.ad, "yemek yiyor.")


class Kedi(Hayvan):
    def miyavla(self):
        print(self.ad, "miyavlıyor.")


class Kopek(Hayvan):
    def havla(self):
        print(self.ad, "havlıyor.")


kedi = Kedi("Pamuk")
kopek = Kopek("Karabaş")

kedi.yemek_ye()
kedi.miyavla()

kopek.yemek_ye()
kopek.havla()
````

Aşağıdakilerden hangileri çalışır?

```python
kedi.havla()
kopek.miyavla()
kedi.yemek_ye()
kopek.yemek_ye()
```

Neden?

### Cevap

```python
kedi.havla()       # ÇALIŞMAZ
kopek.miyavla()    # ÇALIŞMAZ
kedi.yemek_ye()    # ÇALIŞIR
kopek.yemek_ye()   # ÇALIŞIR
```

Çünkü `yemek_ye()` üst sınıftan gelir.

`miyavla()` yalnızca `Kedi`ye, `havla()` yalnızca `Kopek`e özgüdür.

# BÖLÜM 10 — Şimdi UML nedir?

Kod büyüdükçe sınıfların ilişkisini yalnızca kod üzerinden görmek zorlaşabilir.

UML = Unified Modeling Language.

Yazılım sistemlerini görsel olarak ifade etmek için kullanılan modelleme dilidir.

Biz şimdilik sadece **UML Class Diagram (Sınıf Diyagramı)** kullanacağız.

Bir sınıf genellikle üç bölümlü kutuyla gösterilir:

```text
┌──────────────────────────┐
│         Ogrenci          │  ← sınıf adı
├──────────────────────────┤
│ ad                       │
│ email                    │  ← attribute'lar
│ numara                   │
├──────────────────────────┤
│ bilgi_goster()           │  ← metotlar
│ ders_sec()               │
└──────────────────────────┘
```

# BÖLÜM 12 — UML'den kod okuyalım

UML:

```text
              ┌─────────────────┐
              │      Arac       │
              ├─────────────────┤
              │ marka           │
              │ hiz             │
              ├─────────────────┤
              │ hizlan()        │
              └────────▲────────┘
                       │
              ┌────────┴────────┐
              │    Otomobil     │
              ├─────────────────┤
              │ kapi_sayisi     │
              ├─────────────────┤
              │ klima_ac()      │
              └─────────────────┘
```

Önce UML'yi cümleye dönüştürelim:

- Otomobil bir Araçtır.
- Her aracın marka ve hız bilgisi vardır.
- Her araç hızlanabilir.
- Otomobilin ayrıca kapı sayısı vardır.
- Otomobil klima açabilir.

Şimdi kodu yazmak çok daha kolay.

````python
class Arac:
    def __init__(self, marka):
        self.marka = marka
        self.hiz = 0

    def hizlan(self, miktar):
        self.hiz += miktar


class Otomobil(Arac):
    def __init__(self, marka, kapi_sayisi):
        super().__init__(marka)
        self.kapi_sayisi = kapi_sayisi

    def klima_ac(self):
        print("Klima açıldı.")


otomobil = Otomobil("Toyota", 4)

otomobil.hizlan(50)

print(otomobil.marka)
print(otomobil.hiz)
print(otomobil.kapi_sayisi)

otomobil.klima_ac()
````

# BÖLÜM 13 — Koddan UML oluşturalım

Aşağıdaki kodu okuyun:

````python
class Urun:
    def __init__(self, ad, fiyat):
        self.ad = ad
        self.fiyat = fiyat

    def bilgi_goster(self):
        print(self.ad, self.fiyat)


class ElektronikUrun(Urun):
    def __init__(self, ad, fiyat, garanti):
        super().__init__(ad, fiyat)
        self.garanti = garanti

    def garanti_goster(self):
        print("Garanti:", self.garanti, "yıl")
````

## Sıra Sizde 1 — Çözüm

````python
class Telefon:
    def __init__(self, marka, model):
        self.marka = marka
        self.model = model

    def bilgi_goster(self):
        print(self.marka, self.model)


class AkilliTelefon(Telefon):
    def __init__(self, marka, model, isletim_sistemi):
        super().__init__(marka, model)
        self.isletim_sistemi = isletim_sistemi

    def sistem_goster(self):
        print("İşletim sistemi:", self.isletim_sistemi)


telefon = AkilliTelefon(
    "Samsung",
    "S26",
    "Android"
)

telefon.bilgi_goster()
telefon.sistem_goster()
````

# BÖLÜM 15 — HATA BULMA 1: super() unutuldu

Kodu çalıştırmadan tahmin edin:

````python
class Kisi:
    def __init__(self, ad):
        self.ad = ad


class Ogrenci(Kisi):
    def __init__(self, ad, numara):
        self.numara = numara


ogrenci = Ogrenci("Ayşe", 123)

# Bu satır ne olur?
print(ogrenci.ad)
````

### Problem

`Ogrenci` kendi `__init__` metodunu yazdı.

Ama:

```python
Kisi.__init__()
```

çalıştırılmadı.

Dolayısıyla:

```python
self.ad
```

oluşturulmadı.

### Çözüm

````python
class Kisi:
    def __init__(self, ad):
        self.ad = ad


class Ogrenci(Kisi):
    def __init__(self, ad, numara):

        super().__init__(ad)

        self.numara = numara


ogrenci = Ogrenci("Ayşe", 123)

print(ogrenci.ad)
print(ogrenci.numara)
````

# BÖLÜM 18 — HATA BULMA 4: Override sırasında sonsuz çağrı

Aşağıdaki fikir yanlıştır:

```python

    def bilgi_goster(self):
```

Görev:

## Sıra Sizde 2 — Çözüm

````python
class Calisan:
    def __init__(self, ad, maas):
        self.ad = ad
        self.maas = maas

    def zam_yap(self, yuzde):
        self.maas *= 1 + yuzde / 100

    def bilgi_goster(self):
        print(self.ad, "-", self.maas, "TL")


class Yazilimci(Calisan):
    def __init__(self, ad, maas, dil):
        super().__init__(ad, maas)
        self.dil = dil

    def kod_yaz(self):
        print(self.ad, self.dil, "ile kod yazıyor.")


class Yonetici(Calisan):
    def __init__(self, ad, maas, ekip_sayisi):
        super().__init__(ad, maas)
        self.ekip_sayisi = ekip_sayisi

    def toplanti_yap(self):
        print(self.ad, "toplantı yapıyor.")


yazilimci = Yazilimci(
    "Ayşe",
    60000,
    "Python"
)

yonetici = Yonetici(
    "Mehmet",
    80000,
    8
)

yazilimci.zam_yap(10)
yazilimci.bilgi_goster()
yazilimci.kod_yaz()

yonetici.bilgi_goster()
yonetici.toplanti_yap()
````

# BÖLÜM 20 — Çok seviyeli kalıtım

Kalıtım iki seviyede olmak zorunda değildir.

Örneğin:

```text
Canli
  ▲
  │
Hayvan
  ▲
  │
Kedi
```

- Kedi bir Hayvandır.
- Hayvan bir Canlıdır.
- Dolayısıyla Kedi üst zincirdeki davranışları da kullanabilir.

````python
class Canli:
    def nefes_al(self):
        print("Nefes alıyor.")


class Hayvan(Canli):
    def hareket_et(self):
        print("Hareket ediyor.")


class Kedi(Hayvan):
    def miyavla(self):
        print("Miyav!")


kedi = Kedi()

kedi.nefes_al()
kedi.hareket_et()
kedi.miyavla()
````

# BÖLÜM 21 — isinstance() ve issubclass()

Kalıtım ilişkisini Python'a da sorabiliriz.

`isinstance()` → nesne için

`issubclass()` → sınıf için

````python
print(isinstance(kedi, Kedi))
print(isinstance(kedi, Hayvan))
print(isinstance(kedi, Canli))

print()

print(issubclass(Kedi, Hayvan))
print(issubclass(Kedi, Canli))
print(issubclass(Hayvan, Kedi))
````

Dikkat:

Bir `Kedi` nesnesi aynı zamanda `Hayvan` ve `Canli` olarak da kabul edilir.

Bu bize tekrar şu ilişkiyi gösterir:

```text
Kedi bir Hayvandır.
Hayvan bir Canlıdır.
```

# BÖLÜM 26 — HATA BULMA: PyQt

Aşağıdaki kodda ne eksik?

İpucu:

`QMainWindow` nesnesinin kendi başlangıç işlemleri çalıştırıldı mı?

Eksik satır:

```python

```
