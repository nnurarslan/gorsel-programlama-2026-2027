# 7. `__init__` neden gerekli?

Bir önceki yöntemde her nesneyi oluşturduktan sonra tek tek:

```python
ogrenci1.ad = ...
ogrenci1.numara = ...
ogrenci1.notu = ...
```

yazdık.

Bunu daha düzenli yapabilmek için `__init__` kullanabiliriz.

`__init__`, nesne oluşturulduğunda otomatik çalışan özel bir metottur.

````python
class Ogrenci:
    def __init__(self, ad, numara, notu):
        self.ad = ad
        self.numara = numara
        self.notu = notu
````

````python
ogrenci1 = Ogrenci("Ayşe", 101, 85)
ogrenci2 = Ogrenci("Mehmet", 102, 70)

print(ogrenci1.ad)
print(ogrenci2.numara)
````

# 8. `self` nedir?

Başlangıçta en çok karıştırılan kavramlardan biridir.

Basit düşünelim:

`self` = **şu anda üzerinde işlem yapılan nesne**

Şu kod çalıştığında:

```python
ogrenci1 = Ogrenci("Ayşe", 101, 85)
```

`__init__` içerisinde `self`, `ogrenci1` nesnesini temsil eder.

Şu kod çalıştığında:

```python
ogrenci2 = Ogrenci("Mehmet", 102, 70)
```

aynı sınıf kullanılmasına rağmen bu kez `self`, `ogrenci2` nesnesini temsil eder.

````python
class Ogrenci:
    def __init__(self, ad):
        print("Şu anda oluşturulan nesne:", self)
        self.ad = ad

ogrenci1 = Ogrenci("Ayşe")
ogrenci2 = Ogrenci("Mehmet")
````

Şu satırı parçalayalım:

```python
self.ad = ad
```

Sağdaki `ad`:

→ dışarıdan gelen değer

Soldaki `self.ad`:

→ bu değerin nesnenin içerisinde saklandığı attribute

Örneğin:

```python
Ogrenci("Ayşe")
```

çağrıldığında kabaca:

```text
ad = "Ayşe"
self.ad = "Ayşe"
```

olur.

# 10. İlk metodumuzu yazalım

# 11. Metot nesnenin verisini değiştirebilir

````python
class Araba:
    def __init__(self, marka):
        self.marka = marka
        self.hiz = 0

    def hizlan(self):
        self.hiz += 10

araba1 = Araba("Toyota")

print("Başlangıç:", araba1.hiz)

araba1.hizlan()
print("1. hızlanma:", araba1.hiz)

araba1.hizlan()
print("2. hızlanma:", araba1.hiz)
````

Burada `hizlan()` metodu `self.hiz` değerini değiştirdi.

Yani nesne yalnızca veri saklamıyor; kendi durumunu değiştiren davranışlara da sahip olabiliyor.

# 12. Biraz daha gerçekçi Araba sınıfı

````python
class Araba:
    def __init__(self, marka, renk):
        self.marka = marka
        self.renk = renk
        self.hiz = 0

    def hizlan(self, miktar):
        self.hiz += miktar

    def yavasla(self, miktar):
        self.hiz -= miktar

        if self.hiz < 0:
            self.hiz = 0

    def bilgi_goster(self):
        print(f"{self.marka} | {self.renk} | Hız: {self.hiz}")

araba1 = Araba("Toyota", "Beyaz")

araba1.bilgi_goster()
araba1.hizlan(30)
araba1.bilgi_goster()
araba1.yavasla(10)
araba1.bilgi_goster()
````

# 16. Kolay Alıştırma — Dikdörtgen

`Dikdortgen` isimli bir sınıf yazın.

Attribute'lar:

- kısa kenar
- uzun kenar

Metotlar:

- `alan()`
- `cevre()`

Örnek:

```python
d1 = Dikdortgen(5, 10)

print(d1.alan())   # 50
print(d1.cevre())  # 30
```

Önce kendiniz deneyin.

````python
# ÇÖZÜMÜ GÖRMEDEN ÖNCE BURAYA KENDİ KODUNUZU YAZIN
````

## Çözüm — Dikdörtgen

````python
class Dikdortgen:
    def __init__(self, kisa, uzun):
        self.kisa = kisa
        self.uzun = uzun

    def alan(self):
        return self.kisa * self.uzun

    def cevre(self):
        return 2 * (self.kisa + self.uzun)

d1 = Dikdortgen(5, 10)

print("Alan:", d1.alan())
print("Çevre:", d1.cevre())
````

# 18. Orta Seviye — Banka Hesabı

Bir banka hesabının:

- hesap sahibinin adı
- bakiyesi

olsun.

Yapabileceği işlemler:

- para yatırmak
- para çekmek
- bakiye göstermek

Para çekerken bakiye kontrolü de yapalım.

````python
class BankaHesabi:
    def __init__(self, sahip, bakiye=0):
        self.sahip = sahip
        self.bakiye = bakiye

    def para_yatir(self, miktar):
        if miktar > 0:
            self.bakiye += miktar

    def para_cek(self, miktar):
        if miktar <= 0:
            print("Geçersiz miktar.")
        elif miktar > self.bakiye:
            print("Yetersiz bakiye.")
        else:
            self.bakiye -= miktar

    def bilgi_goster(self):
        print(f"{self.sahip} | Bakiye: {self.bakiye} TL")

hesap = BankaHesabi("Ayşe", 1000)

hesap.bilgi_goster()
hesap.para_yatir(500)
hesap.bilgi_goster()
hesap.para_cek(300)
hesap.bilgi_goster()
````

# 19. Orta Seviye Alıştırma — Öğrenci Not Sistemi

`Ogrenci` sınıfı oluşturun.

Attribute'lar:

- ad
- vize
- final

Metotlar:

- `ortalama()`
- `durum()`

Ortalama:

`vize × 0.40 + final × 0.60`

Ortalama 50 veya üzeriyse `Geçti`, değilse `Kaldı`.

````python
# Buraya çözümünüzü yazın.
````

## Çözüm — Öğrenci Not Sistemi

````python
class Ogrenci:
    def __init__(self, ad, vize, final):
        self.ad = ad
        self.vize = vize
        self.final = final

    def ortalama(self):
        return self.vize * 0.40 + self.final * 0.60

    def durum(self):
        if self.ortalama() >= 50:
            return "Geçti"
        return "Kaldı"

ogrenci1 = Ogrenci("Ayşe", 70, 80)
ogrenci2 = Ogrenci("Mehmet", 30, 40)

print(ogrenci1.ad, ogrenci1.ortalama(), ogrenci1.durum())
print(ogrenci2.ad, ogrenci2.ortalama(), ogrenci2.durum())
````

# 21. Zor Örnek — Stok Yönetimi

Bir `Urun` sınıfı tasarlayalım.

Attribute'lar:

- ad
- fiyat
- stok

Metotlar:

- stok ekle
- satış yap
- zam yap
- bilgi göster

Satış yapılırken yeterli stok olup olmadığı kontrol edilsin.

````python
class Urun:
    def __init__(self, ad, fiyat, stok):
        self.ad = ad
        self.fiyat = fiyat
        self.stok = stok

    def stok_ekle(self, adet):
        if adet > 0:
            self.stok += adet

    def satis_yap(self, adet):
        if adet <= 0:
            print("Geçersiz adet.")
        elif adet > self.stok:
            print("Yeterli stok yok.")
        else:
            self.stok -= adet
            print(f"{adet} adet satış yapıldı.")

    def zam_yap(self, yuzde):
        if yuzde > 0:
            self.fiyat *= 1 + yuzde / 100

    def bilgi_goster(self):
        print(
            f"Ürün: {self.ad} | "
            f"Fiyat: {self.fiyat:.2f} TL | "
            f"Stok: {self.stok}"
        )

urun = Urun("Klavye", 1000, 5)

urun.bilgi_goster()
urun.satis_yap(2)
urun.zam_yap(10)
urun.stok_ekle(3)
urun.bilgi_goster()
````

# 22. Zor Alıştırma — Kütüphane Kitabı

`Kitap` sınıfı oluşturun.

Attribute'lar:

- ad
- yazar
- odunc_alindi (`False` ile başlasın)

Metotlar:

### odunc_al()

Kitap müsaitse ödünç verilsin.

Zaten ödünç alınmışsa:

`Kitap şu anda müsait değil.`

mesajı gösterilsin.

### iade_et()

Kitap tekrar müsait hale gelsin.

### bilgi_goster()

Kitabın adı, yazarı ve durumu gösterilsin.

````python
# Buraya çözümünüzü yazın.
````

## Çözüm — Kütüphane Kitabı

````python
class Kitap:
    def __init__(self, ad, yazar):
        self.ad = ad
        self.yazar = yazar
        self.odunc_alindi = False

    def odunc_al(self):
        if self.odunc_alindi:
            print("Kitap şu anda müsait değil.")
        else:
            self.odunc_alindi = True
            print("Kitap ödünç verildi.")

    def iade_et(self):
        if not self.odunc_alindi:
            print("Kitap zaten kütüphanede.")
        else:
            self.odunc_alindi = False
            print("Kitap iade edildi.")

    def bilgi_goster(self):
        durum = "Ödünçte" if self.odunc_alindi else "Müsait"
        print(f"{self.ad} - {self.yazar} | {durum}")

kitap1 = Kitap("1984", "George Orwell")

kitap1.bilgi_goster()
kitap1.odunc_al()
kitap1.bilgi_goster()
kitap1.odunc_al()
kitap1.iade_et()
kitap1.bilgi_goster()
````

# 23. Daha Zor — Sepet ve Ürün Nesneleri

Şimdi nesnelerin birbirleriyle çalışmasına bakalım.

Bir `Urun` nesnesini başka bir nesnenin içinde kullanabiliriz.

````python
class Urun:
    def __init__(self, ad, fiyat):
        self.ad = ad
        self.fiyat = fiyat


class Sepet:
    def __init__(self):
        self.urunler = []

    def urun_ekle(self, urun):
        self.urunler.append(urun)

    def toplam(self):
        toplam_fiyat = 0

        for urun in self.urunler:
            toplam_fiyat += urun.fiyat

        return toplam_fiyat

    def sepeti_goster(self):
        for urun in self.urunler:
            print(urun.ad, "-", urun.fiyat, "TL")

        print("Toplam:", self.toplam(), "TL")


klavye = Urun("Klavye", 1000)
mouse = Urun("Mouse", 500)
monitor = Urun("Monitör", 6000)

sepet = Sepet()

sepet.urun_ekle(klavye)
sepet.urun_ekle(mouse)
sepet.urun_ekle(monitor)

sepet.sepeti_goster()
````

Burada önemli bir aşamaya geldik:

`Sepet` nesnesinin içerisinde `Urun` nesneleri bulunmaktadır.

Yani nesneler yalnızca kendi başlarına kullanılmaz; başka nesnelerle birlikte bir sistem oluşturabilirler.

Gerçek yazılımlarda OOP'nin gücü burada daha belirgin hale gelir.

## Hata Bulma — Açıklama

İlk parametrenin adı teknik olarak `self` olmak zorunda değildir.

Ancak Python'da standart kullanım `self` şeklindedir.

Asıl sorun sınıfın tasarımının karışık yazılmasıdır. Doğru ve okunabilir biçim:

```python
class Araba:
    def __init__(self, marka, renk):
        self.marka = marka
        self.renk = renk
```

Burada:

- `self` → oluşturulan nesne
- `marka` → dışarıdan gelen marka
- `renk` → dışarıdan gelen renk

````python
class Bilgisayar:
    def __init__(self, marka, ram):
        self.marka = marka
        self.ram = ram

    def ram_artir(self, miktar):
        self.ram += miktar

pc1 = Bilgisayar("Lenovo", 16)
pc2 = Bilgisayar("Asus", 8)

pc1.ram_artir(8)
````

`Personel` sınıfı oluşturun.

Metotlar:

### bilgi_goster()

### zam_yap(yuzde)

### departman_degistir(yeni_departman)

Ardından en az 3 farklı personel nesnesi oluşturun.

Personelleri bir listede tutun ve tüm personellerin bilgilerini `for` döngüsüyle yazdırın.

Bonus:

Maaşı 50.000 TL'den fazla olan personelleri ayrıca listeleyin.

````python
# FINAL GÖREVİ ÇÖZÜM ALANI
````

## Final Görevi — Örnek Çözüm

````python
class Personel:
    def __init__(self, ad, departman, maas):
        self.ad = ad
        self.departman = departman
        self.maas = maas

    def bilgi_goster(self):
        print(
            f"{self.ad} | "
            f"{self.departman} | "
            f"{self.maas:.2f} TL"
        )

    def zam_yap(self, yuzde):
        if yuzde > 0:
            self.maas *= 1 + yuzde / 100

    def departman_degistir(self, yeni_departman):
        self.departman = yeni_departman


personeller = [
    Personel("Ayşe", "Yazılım", 55000),
    Personel("Mehmet", "Muhasebe", 42000),
    Personel("Zeynep", "Yazılım", 60000)
]

print("Tüm personeller:")
for personel in personeller:
    personel.bilgi_goster()

print("\n50.000 TL üzeri:")
for personel in personeller:
    if personel.maas > 50000:
        personel.bilgi_goster()
````
