## 4. Sayılar ve Operatörler

Şimdi Python'ın temel aritmetik operatörlerini hatırlayalım.

`+  -  *  /  //  %  **`


````python
a = 17
b = 4

print("Toplama:", a + b)
print("Çıkarma:", a - b)
print("Çarpma:", a * b)
print("Bölme:", a / b)
print("Tam bölme:", a // b)
print("Kalan (mod):", a % b)
print("Üs:", a ** b)
````

### 💬 Sınıf Sorusu

Aşağıdaki üç işlemin farkı nedir?

- `17 / 4`
- `17 // 4`
- `17 % 4`


````python
print(17 / 4)
print(17 // 4)
print(17 % 4)
````

## 5. Hazır Modüller — `math` ve `random`

````python
import math

print("16'nın karekökü:", math.sqrt(16))
print("5! =", math.factorial(5))
print("Pi:", math.pi)
````

## 6. Kullanıcıdan Veri Alma — `input()`

`input()` ile kullanıcıdan veri alabiliriz. Ancak önemli bir ayrıntı var: `input()` sonucu varsayılan olarak **string** gelir.


````python
sayi = input("Bir sayı giriniz: ")

print("Girilen değer:", sayi)
print("Veri tipi:", type(sayi))
````

### Tip dönüşümü

Matematiksel işlem yapacaksak girdiyi uygun veri tipine dönüştürmeliyiz.


````python
sayi = int(input("Bir tam sayı giriniz: "))

print("Sayının toplamı:", sayi + 2)
print("Veri tipi:", type(sayi))
````

# 🛒 Ders Sonu Mini Uygulaması — Alışveriş Sepeti

Şimdi öğrendiğimiz konuları tek bir küçük programda birleştirelim.

Program kullanıcıdan:
1. Ürün adını,
2. Ürün fiyatını,
3. Adedi

alacak ve toplam tutarı hesaplayacak.


<details>
<summary><strong>💡 Çözümü Göster</strong></summary>

> Önce kendi çözümünüzü yazmayı deneyin.

```python
urun = input("Ürün adı: ")
fiyat = float(input("Ürün fiyatı: "))
adet = int(input("Adet: "))

toplam = fiyat * adet

print("\n------ FİŞ ------")
print(f"Ürün   : {urun}")
print(f"Fiyat  : {fiyat:.2f} TL")
print(f"Adet   : {adet}")
print(f"Toplam : {toplam:.2f} TL")
print("-----------------")
```

</details>
