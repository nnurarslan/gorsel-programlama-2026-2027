## 1. Listeler (`list`)

Liste, birden fazla değeri **sıralı** biçimde tutmamızı sağlar.  
Listeler **değiştirilebilir (mutable)** yapılardır ve aynı değeri birden fazla kez içerebilir.

```text
indeks:     0        1        2       3
         "elma"   "armut"   "kivi"   "muz"
```

Negatif indeksler sondan başlar: `-1` son elemandır.


````python
bos_liste = []
print(type(bos_liste))
````

````python
meyveler = ["elma", "armut", "kivi", "muz"]
print(meyveler)
````

### Listeye eleman ekleme

- `append()` → sona tek eleman ekler.
- `insert()` → belirtilen indekse ekler.
- `extend()` → başka bir koleksiyonun elemanlarını listeye ekler.


````python
meyveler.append("çilek")
print(meyveler)

meyveler.insert(2, "şeftali")
print(meyveler)

yeni_meyveler = ["kavun", "karpuz"]
meyveler.extend(yeni_meyveler)
print(meyveler)
````

### Listeleri birleştirme

`+` operatörü yeni bir liste oluşturur. Sonucu saklamak istiyorsak bir değişkene atamalıyız.


````python
yeni = ["ejder_meyvesi"]
print(meyveler + yeni)
print("Eski liste:", meyveler)
````

````python
meyveler = meyveler + yeni
print("Yeni liste:", meyveler)
````

### İndeksleme ve dilimleme (`slicing`)


````python
print("3. eleman:", meyveler[2])
print("Son eleman:", meyveler[-1])
print("2:5 arası:", meyveler[2:5])
print("2. indeksten sonrası:", meyveler[2:])
print("Sondan dilim:", meyveler[-3:-1])
````

### Eleman değiştirme ve silme

Listeler değiştirilebilir olduğu için bir indeksteki değeri doğrudan değiştirebiliriz.


````python
meyveler[1] = "greyfurt"
print(meyveler)
````

````python
meyveler.remove("şeftali")   # Değere göre siler
print(meyveler)

````

````python

silinen = meyveler.pop()       # Varsayılan olarak son elemanı siler ve döndürür
print("Silinen:", silinen)
print(meyveler)
````

### Liste metotları


````python
sayilar = [1, 2, 3, 4, 5, 2, 2]

print("Uzunluk:", len(sayilar))
print("2 kaç kez var?:", sayilar.count(2))
print("2 ilk hangi indekste?:", sayilar.index(2))
print("İkişer adımlı:", sayilar[::2])
````

````python
sayilar.sort()
print("Küçükten büyüğe:", sayilar)

sayilar.sort(reverse=True)
print("Büyükten küçüğe:", sayilar)

sayilar.reverse()
print("Ters çevrilmiş:", sayilar)
````

### Liste kopyalama

`copy()` veya `[:]` kullanarak listenin ayrı bir kopyasını oluşturabiliriz.


````python
kopya1 = meyveler.copy()
kopya2 = meyveler[:]
 
print(kopya1)
print(kopya2)
````

## 2. Sözlükler (`dict`)

Sözlükler veriyi **anahtar : değer (key : value)** çiftleri halinde saklar.

Bir öğrenciyi yalnızca bir listeyle tutmak yerine:

```python
{"ad": "Ayşe", "yas": 20, "ortalama": 3.25}
```

şeklinde anlamlı alan adlarıyla temsil edebiliriz.


````python
kisi_bilgileri = {
    "ad": "Nur",
    "meslek": "mühendis",
    "yas": 29
}

print(kisi_bilgileri)
print("Meslek:", kisi_bilgileri["meslek"])
````

### Yeni alan ekleme, güncelleme ve silme


````python
kisi_bilgileri["sehir"] = "Kütahya"
kisi_bilgileri["yas"] = 30
print(kisi_bilgileri)


````

````python
del kisi_bilgileri["meslek"]
print(kisi_bilgileri)
````

### Anahtarlar, değerler ve çiftler


````python
print(kisi_bilgileri.items())
````

````python
print("Anahtarlar:", kisi_bilgileri.keys())
print("Değerler:", kisi_bilgileri.values())
print("Çiftler:", kisi_bilgileri.items())
````

### İç içe sözlükler

Birden fazla kişiyi tek bir sözlük içerisinde ayrı kayıtlar olarak tutabiliriz.


````python
kisiler = {
    "kisi_1": {"ad": "Ayşe", "yas": 26},
    "kisi_2": {"ad": "Ahmet", "yas": 27}
}

print(kisiler["kisi_2"]["yas"])
print(kisiler["kisi_1"]["ad"])
````

## 3. Demetler (`tuple`)

Tuple, listeye benzeyen **sıralı** bir veri yapısıdır; ancak oluşturulduktan sonra elemanları doğrudan değiştirilemez (**immutable**).

Sabit kalmasını istediğimiz veri gruplarında kullanışlıdır.


````python
my_tuple = (1, 2, 3, 4)
single_element_tuple = (5,)

print(my_tuple)
print(type(single_element_tuple))
print(my_tuple[1:3])
````

````python
print("Toplam:", sum(my_tuple))
print("Minimum:", min(my_tuple))
print("Maksimum:", max(my_tuple))
````

### Tuple birleştirme ve üyelik kontrolü


````python
tuple1 = (1, 2)
tuple2 = (3, 4)
result = tuple1 + tuple2

print(result)
print(5 in result)
print(len(result))
````

````python
result[1] = 5
````

### ⚠️ Tuple neden farklı?

Aşağıdaki ifade hata üretir:

```python
result[1] = 5
```

Çünkü tuple elemanları doğrudan değiştirilemez. Değişiklik gerekiyorsa geçici olarak listeye çevrilebilir.


````python
my_tuple = (1, 2, 3, 4)
my_list = list(my_tuple)
my_list[1] = 20
my_tuple = tuple(my_list)
print(my_tuple)
````

### Tuple unpacking

Tuple elemanlarını birden fazla değişkene tek satırda dağıtabiliriz.


````python
my_tuple = (1, 2, 3)
a, b, c = my_tuple
print(a, b, c)
````

````python
my_tuple = (1, 2, 3, 4, 5)
a, b, *rest = my_tuple
print("a:", a)
print("b:", b)
print("rest:", rest)
````

````python
my_tuple = (1, 2, 3, 4, 5)
a, *middle, d, e = my_tuple
print("middle:", middle)
print("d:", d, "e:", e)
````

## 4. Kümeler (`set`)

Set, **benzersiz** elemanlardan oluşan bir koleksiyondur. Aynı değer birden fazla kez eklenirse tekrar oluşturmaz.

> Set'lerde indeks mantığı yoktur; elemanların belirli bir sırada tutulacağı varsayılmamalıdır.


````python
my_set = {1, 2, 3}
my_set.add(4)

````

````python
print(my_set)
````

````python
my_set.add(4)
print(my_set)
````

### Eleman silme

- `remove(x)` → eleman yoksa hata verir.
- `discard(x)` → eleman yoksa hata vermez.
- `pop()` → bir elemanı kaldırır; hangi elemanın geleceğine güvenilmemelidir.


````python
my_set = {1, 2, 3, 4}
my_set.remove(3)
print(my_set)
````

````python
my_set.remove(3)
print(my_set)
````

````python
my_set.discard(5)
print(my_set)
````

### Küme işlemleri


````python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

print("Birleşim:", set1.union(set2))
print("Birleşim | :", set1 | set2)

print("Kesişim:", set1.intersection(set2))
print("Kesişim & :", set1 & set2)

print("Fark:", set1.difference(set2))
print("Fark - :", set1 - set2)
````

# 🎯 Ders Sonu Uygulaması — Öğrenci Veri Modeli

Şimdi dört veri yapısını **aynı problem içerisinde** kullanalım.

`ogrenciler` isminde bir dictionary oluşturun ve iki öğrenci ekleyin.

Her öğrencide şu alanlar bulunsun:

- `ad` → `str`
- `yas` → `int`
- `notlar` → `list`
- `dersler` → `set`
- `gezdigi_yerler` → `tuple`

Ardından:

1. Bir öğrencinin adını yazdırın.
2. Bir öğrencinin not ortalamasını hesaplayın.
3. Öğrencinin aldığı dersleri gösterin.
4. Gezdiği yerlerden ilkini yazdırın.
5. Öğrencinin derslerine `"Görsel Programlama"` dersini ekleyin.


````python
# Çözümünüzü buraya yazın
ogrenciler = {
    "ogrenci_1": {
        "ad": "Ayşe",
        "yas": 20,
        "notlar": [75, 85, 90],
        "dersler": {"Python", "Matematik"},
        "gezdigi_yerler": ("Kütahya", "Eskişehir", "İstanbul")
    },
    "ogrenci_2": {
        "ad": "Ahmet",
        "yas": 21,
        "notlar": [60, 70, 80],
        "dersler": {"Python", "Fizik"},
        "gezdigi_yerler": ("Ankara", "İzmir")
    }
}

print("Ad:", ogrenciler["ogrenci_1"]["ad"])

notlar = ogrenciler["ogrenci_1"]["notlar"]
ortalama = sum(notlar) / len(notlar)
print("Not ortalaması:", ortalama)

print("Dersler:", ogrenciler["ogrenci_1"]["dersler"])
print("İlk gezdiği yer:", ogrenciler["ogrenci_1"]["gezdigi_yerler"][0])

ogrenciler["ogrenci_1"]["dersler"].add("Görsel Programlama")
print("Güncel dersler:", ogrenciler["ogrenci_1"]["dersler"])
````

| Veri Yapısı | Sıralı | Değiştirilebilir | Tekrar | Tipik kullanım |

| `tuple` | Evet | Hayır | Var | Sabit veri grupları, koordinatlar |
| `set` | Sıraya güvenilmez | Evet | Yok | Benzersiz değerler, küme işlemleri |
