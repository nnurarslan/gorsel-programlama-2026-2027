## 1. Fonksiyon Nedir ve Neden Kullanılır?

Fonksiyon, belirli bir işi yapmak üzere isimlendirilmiş ve gerektiğinde tekrar çağrılabilen kod bloğudur.

Fonksiyon kullanmak:

- aynı kodu tekrar tekrar yazmayı önler,
- programı daha okunabilir hâle getirir,
- büyük problemleri küçük parçalara ayırır,
- test ve hata ayıklamayı kolaylaştırır,
- GUI uygulamalarında her işlemin ayrı bir görev olarak tasarlanmasını sağlar.

Bir fonksiyon `def` anahtar kelimesiyle tanımlanır.


````python
def merhaba():
    print("Merhaba Dünya")

merhaba()
merhaba()
merhaba()
````

`merhaba()` fonksiyonunun parametresi yoktur ve geriye bir değer döndürmez. Her çağrıldığında içindeki komutlar yeniden çalışır.

## 2. Parametre ve Argüman

Fonksiyon tanımında kullanılan değişkenlere **parametre**, fonksiyon çağrılırken gönderilen gerçek değerlere **argüman** denir.

Aşağıdaki fonksiyonda `a` ve `b` parametredir; `5` ve `10` ise argümandır.


````python
def toplam(a, b):
    return a + b

sonuc = toplam(5, 10)
print(sonuc)
````

## 3. `return` Neden Önemlidir?

`print()` yalnızca bir değeri ekrana gösterir. `return` ise fonksiyonun ürettiği sonucu çağrıldığı yere **geri gönderir**.

Bu nedenle `return` edilen sonuç:
- bir değişkene atanabilir,
- başka bir hesaplamada kullanılabilir,
- başka bir fonksiyona gönderilebilir.

GUI uygulamalarında hesaplama fonksiyonunun sonucu `return` etmesi, sonucu daha sonra bir `QLabel` üzerinde gösterebilmemizi sağlar.


````python
def kare(sayi):
    return sayi ** 2

sonuc = kare(6)
print("Sonuç:", sonuc)
print("Sonucun 10 fazlası:", sonuc + 10)
````

## 8. `*args` — Değişken Sayıda Konumsal Argüman

Kaç tane argüman geleceğini önceden bilmiyorsak `*args` kullanılabilir. Gelen değerler fonksiyon içinde tuple benzeri bir yapı olarak ele alınır.


````python
def toplama(*args):
    return sum(args)

print(toplama(1, 2))
print(toplama(1, 2, 3, 4))
print(toplama(10, 20, 30, 40, 50))
````

## 9. `**kwargs` — Değişken Sayıda İsimlendirilmiş Argüman

`**kwargs`, `anahtar=değer` biçiminde gönderilen değişken sayıdaki bilgiyi dictionary mantığıyla alır.


````python
def kisi_bilgileri(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

kisi_bilgileri(isim="Mehmet", yas=28, sehir="İstanbul")
````

## 10. Değişken Kapsamı (Scope)

Bir değişkenin programın hangi bölümünden erişilebilir olduğunu **scope** belirler.

- Fonksiyon dışında oluşturulan değişken genellikle **global** kapsamlıdır.
- Fonksiyon içinde oluşturulan değişken **local** kapsamlıdır.
- Aynı isimde local ve global değişken bulunabilir.


````python
x = 30

def func():
    x = 20
    print("Fonksiyon içindeki x:", x)

func()
print("Fonksiyon dışındaki x:", x)
````

Fonksiyon içindeki `x`, dışarıdaki `x`'ten farklı bir local değişkendir.

### `global` Anahtar Kelimesi

Bir fonksiyon içinde global değişkeni değiştirmek mümkündür; ancak gereksiz `global` kullanımı programın takibini zorlaştırır. Mümkün olduğunda değeri parametre olarak almak ve sonucu `return` etmek daha temiz bir tasarımdır.


````python
kur = 34.5

def kuru_degistir():
    global kur
    kur = 35.0
    print(kur)

kuru_degistir()
print(kur)
````

### Local Değişkene Dışarıdan Erişilemez

Aşağıdaki örnekte `y`, yalnızca fonksiyon içinde vardır. `print(y)` satırı fonksiyon dışında çalıştırılırsa `NameError` oluşur.


````python
def func():
    y = 20
    print("Fonksiyon içinde:", y)

func()

# Aşağıdaki satır hata verir:
print(y)
````

## 11. İç İçe Fonksiyonlar

Bir fonksiyonun içinde başka bir fonksiyon tanımlanabilir. İç fonksiyon, dış fonksiyonun local değişkenlerine erişebilir.


````python
def dis_fonksiyon():
    mesaj = "Merhaba"

    def ic_fonksiyon():
        print(mesaj)

    ic_fonksiyon()

dis_fonksiyon()
````

## Sıra Sizde 1 — Not Analiz Fonksiyonu

`not_analiz(notlar)` isimli bir fonksiyon yazın.

Fonksiyon bir not listesi almalı ve:
- ortalamayı,
- en yüksek notu,
- en düşük notu,
- 50 ve üzeri not alan öğrenci sayısını

hesaplayarak **4 değer birden `return` etmelidir**.

Örnek veri:

```python
notlar = [45, 78, 90, 32, 67, 55, 100]
```


````python
# Çözümünüzü buraya yazın

def not_analiz(notlar):
    ortalama = sum(notlar) / len(notlar)
    en_yuksek = max(notlar)
    en_dusuk = min(notlar)

    gecen_sayisi = 0
    for notu in notlar:
        if notu >= 50:
            gecen_sayisi += 1

    return ortalama, en_yuksek, en_dusuk, gecen_sayisi


notlar = [45, 78, 90, 32, 67, 55, 100]

ortalama, maksimum, minimum, gecen = not_analiz(notlar)

print(f"Ortalama: {ortalama:.2f}")
print("En yüksek:", maksimum)
print("En düşük:", minimum)
print("Geçen öğrenci sayısı:", gecen)
````

## Sıra Sizde 2 — Gelişmiş Etkinlik Ücreti

`etkinlik_ucreti(yas, ogrenci, gelir, normal_ucret=100)` isimli bir fonksiyon yazın.

Kurallar:
- 18 yaşından küçük → %50 indirim
- 18–25 yaş arasında **ve öğrenci** → %30 indirim
- Geliri 15.000 TL'nin altında olan ve yukarıdaki indirimlerden yararlanmayan kullanıcı → %10 indirim
- Diğer durumlar → normal ücret

Fonksiyon **ödenecek ücret ile uygulanan indirim açıklamasını birlikte döndürsün**.

> Burada koşulların sırası önemlidir. Bir kullanıcı yalnızca ilk sağladığı indirimden yararlansın.


````python
# Çözümünüzü buraya yazın

def etkinlik_ucreti(yas, ogrenci, gelir, normal_ucret=100):
    if yas < 18:
        return normal_ucret * 0.50, "%50 yaş indirimi"

    elif 18 <= yas <= 25 and ogrenci.lower() == "evet":
        return normal_ucret * 0.70, "%30 öğrenci indirimi"

    elif gelir < 15000:
        return normal_ucret * 0.90, "%10 gelir indirimi"

    return normal_ucret, "İndirim uygulanmadı"


yas = int(input("Yaş: "))
ogrenci = input("Öğrenci misiniz? (evet/hayır): ")
gelir = float(input("Gelir: "))

ucret, aciklama = etkinlik_ucreti(yas, ogrenci, gelir)

print(aciklama)
print(f"Ödenecek ücret: {ucret:.2f} TL")
````

## Sıra Sizde 4 — Öğrenci Raporu (`**kwargs`)

`ogrenci_raporu(**bilgiler)` isimli bir fonksiyon yazın.

Fonksiyona şu şekilde farklı bilgiler gönderilebilsin:

```python
ogrenci_raporu(
    ad="Ayşe",
    numara="2026001",
    bolum="Yazılım Mühendisliği",
    vize=75,
    final=85
)
```

Fonksiyon:
- gönderilen bilgileri ekrana yazdırsın,
- `vize` ve `final` varsa başarı notunu `%40 vize + %60 final` ile hesaplasın,
- başarı notu 50 ve üzerindeyse `Geçti`, değilse `Kaldı` yazdırsın.


````python
# Çözümünüzü buraya yazın
def ogrenci_raporu(**bilgiler):
    for anahtar, deger in bilgiler.items():
        print(f"{anahtar.capitalize()}: {deger}")

    if "vize" in bilgiler and "final" in bilgiler:
        basari = bilgiler["vize"] * 0.40 + bilgiler["final"] * 0.60

        if basari >= 50:
            durum = "Geçti"
        else:
            durum = "Kaldı"

        print(f"Başarı notu: {basari:.2f}")
        print("Durum:", durum)


ogrenci_raporu(
    ad="Ayşe",
    numara="2026001",
    bolum="Yazılım Mühendisliği",
    vize=75,
    final=85
)
````

```python
import math


def dikdortgen_alan(kisa, uzun):
    return kisa * uzun


def dikdortgen_cevre(kisa, uzun):
    return 2 * (kisa + uzun)


def kare_alan(kenar):
    return kenar ** 2


def kare_cevre(kenar):
    return 4 * kenar


def daire_alan(yaricap):
    return math.pi * yaricap ** 2


def daire_cevre(yaricap):
    return 2 * math.pi * yaricap


def main():
    print("1 - Dikdörtgen")
    print("2 - Daire")
    print("3 - Kare")

    secim = input("Seçiminiz: ")

    if secim == "1":
        kisa = float(input("Kısa kenar: "))
        uzun = float(input("Uzun kenar: "))
        alan = dikdortgen_alan(kisa, uzun)
        cevre = dikdortgen_cevre(kisa, uzun)

    elif secim == "2":
        yaricap = float(input("Yarıçap: "))
        alan = daire_alan(yaricap)
        cevre = daire_cevre(yaricap)

    elif secim == "3":
        kenar = float(input("Kenar: "))
        alan = kare_alan(kenar)
        cevre = kare_cevre(kenar)

    else:
        print("Geçersiz seçim.")
        return

    print(f"Alan: {alan:.2f}")
    print(f"Çevre: {cevre:.2f}")


if __name__ == "__main__":
    main()
```

## Sıra Sizde 6 — Zor: ATM İşlem Simülasyonu

Aşağıdaki fonksiyonları kullanarak küçük bir ATM uygulaması geliştirin:

```python
para_yatir(bakiye, miktar)
para_cek(bakiye, miktar)
bakiye_goster(bakiye)
main()
```

Kurallar:

- Başlangıç bakiyesi `5000 TL` olsun.
- Para yatırma miktarı `0` veya negatif olamaz.
- Para çekme miktarı `0` veya negatif olamaz.
- Bakiyeden fazla para çekilemez.
- Her işlem fonksiyon tarafından kontrol edilmelidir.
- Para yatırma ve çekme fonksiyonları **güncellenmiş bakiyeyi ve işlem mesajını `return` etmelidir**.
- Kullanıcı `4 - Çıkış` seçeneğini seçene kadar menü tekrar gösterilmelidir.

**Ek zorluk:** Yapılan işlemleri bir `islem_gecmisi` listesinde tutup çıkışta listeleyin.


````python
# Çözümünüzü buraya yazın

````

<details>
<summary><b>💡 Çözümü Göster</b></summary>

```python
def para_yatir(bakiye, miktar):
    if miktar <= 0:
        return bakiye, "Geçersiz miktar."

    bakiye += miktar
    return bakiye, f"{miktar:.2f} TL yatırıldı."


def para_cek(bakiye, miktar):
    if miktar <= 0:
        return bakiye, "Geçersiz miktar."

    if miktar > bakiye:
        return bakiye, "Yetersiz bakiye."

    bakiye -= miktar
    return bakiye, f"{miktar:.2f} TL çekildi."


def bakiye_goster(bakiye):
    return f"Güncel bakiye: {bakiye:.2f} TL"


def main():
    bakiye = 5000
    islem_gecmisi = []

    while True:
        print("\n1 - Para Yatır")
        print("2 - Para Çek")
        print("3 - Bakiye Göster")
        print("4 - Çıkış")

        secim = input("Seçiminiz: ")

        if secim == "1":
            miktar = float(input("Miktar: "))
            bakiye, mesaj = para_yatir(bakiye, miktar)
            islem_gecmisi.append(mesaj)
            print(mesaj)

        elif secim == "2":
            miktar = float(input("Miktar: "))
            bakiye, mesaj = para_cek(bakiye, miktar)
            islem_gecmisi.append(mesaj)
            print(mesaj)

        elif secim == "3":
            print(bakiye_goster(bakiye))

        elif secim == "4":
            break

        else:
            print("Geçersiz seçim.")

    print("\nİşlem Geçmişi")
    for islem in islem_gecmisi:
        print("-", islem)

    print(bakiye_goster(bakiye))


main()
```

**Açıklama:** Bakiyeyi `global` yapmak yerine fonksiyonlara parametre olarak gönderip güncel değeri `return` ediyoruz. Bu, fonksiyonların bağımlılıklarını azaltır ve daha güvenli bir tasarım oluşturur.

</details>

```python
    if fiyat < 0 or adet <= 0:
        return False

    sepet.append({
        "urun": urun_adi,

        "adet": adet
    })

def ara_toplam_hesapla(sepet):
    toplam = 0

    for urun in sepet:
        toplam += urun["fiyat"] * urun["adet"]

    return toplam


def indirim_hesapla(ara_toplam, ogrenci=False):
    indirim = 0

    if ara_toplam >= 500:
        indirim += ara_toplam * 0.10

    if ogrenci:
        indirim += ara_toplam * 0.05

    return indirim


def kdv_hesapla(tutar, oran=20):
    return tutar * oran / 100


def fis_olustur(sepet, ogrenci=False):
    ara_toplam = ara_toplam_hesapla(sepet)
    indirim = indirim_hesapla(ara_toplam, ogrenci)
    indirimli_tutar = ara_toplam - indirim
```
