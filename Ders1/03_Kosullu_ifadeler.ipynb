## 1. Basit `if`

`if`, verilen koşul `True` olduğunda altındaki girintili kod bloğunu çalıştırır. `sayi` değerini 8 yapıp tekrar deneyin.


````python
sayi = 12
if sayi > 10:
    print("Sayı 10'dan büyüktür.")
    print("test 1")
````

## 2. `if - else`

`else`, `if` koşulu sağlanmadığında çalışır.


````python
sayi = 9
if sayi > 10:
    print("Sayı 10'dan büyüktür.")
else:
    print("Sayı 10'dan küçük veya eşittir.")
````

## 3. `if - elif - else`

`elif`, birden fazla koşulu sırayla kontrol etmemizi sağlar. İlk doğru koşulun bloğu çalışır.


````python
sayi = 6
if sayi > 10:
    print("Sayı 10'dan büyüktür.")
elif sayi > 5:
    print("Sayı 5'ten büyük, 10'dan küçük veya eşittir.")
else:
    print("Sayı 5 veya daha küçüktür.")
````

## 4. Tek / Çift Kontrolü

`%` kalan operatörüdür. Bir sayının 2 ile bölümünden kalan 0 ise sayı çifttir.


````python
sayi = 6
if sayi % 2 == 0:
    print("Çift sayı")
else:
    print("Tek sayı")
````

## 5. Pozitif / Negatif / Sıfır

Üç olası durum `if`, `elif` ve `else` ile ayrılır.


````python
sayi = 0
if sayi > 0:
    print("Sayı pozitiftir.")
elif sayi < 0:
    print("Sayı negatiftir.")
else:
    print("Sayı sıfırdır.")
````

## 6. Tek Satırlık `if`

Kısa işlemler tek satırda yazılabilir; okunabilirlik için normal blok yapısı çoğu zaman daha uygundur.


````python
a = 5
b = 3
if a > b: print("a, b'den büyüktür.")
````

## 7. Koşullu İfade (Ternary)

Python'da basit iki durumlu seçimler tek satırlık koşullu ifadeyle yazılabilir.


````python
a = 5
b = 3
print("a büyüktür") if a > b else print("b büyük veya eşittir")
````

## 8. Aralık Kontrolü

Python zincirleme karşılaştırmayı destekler. Bu ifade 10 ≤ sayı ≤ 20 anlamına gelir.


````python
sayi = 15
if 10 <= sayi <= 20:
    print("Sayı 10 ile 20 arasında.")
else:
    print("10-20 aralığında değil.")
````

## 9. `or` Operatörü

`or` kullanıldığında koşullardan en az birinin doğru olması yeterlidir.


````python
gun = "pazartesi"
if gun == "cumartesi" or gun == "pazar":
    print("Hafta sonu")
else:
    print("Hafta içi")
````

## 10. `and` Operatörü

`and` kullanıldığında iki koşulun da doğru olması gerekir. Boolean değişken doğrudan koşul olarak kullanılabilir.


````python
yas = 25
ehliyet = False
if yas >= 18 and ehliyet:
    print("Araç sürebilir.")
else:
    print("Araç süremez.")
````

## 11. İç İçe `if`

Bir koşulun sonucu içinde başka bir koşul kontrol edilebilir. Buna iç içe (nested) `if` denir.


````python
yas = 18
ehliyet = False
if yas >= 18:
    if ehliyet:
        print("Araç sürebilir.")
    else:
        print("Ehliyetin yok, araç süremezsin.")
else:
    print("Yaş şartı sağlanmıyor.")
````

## 12. Ternary ile Değer Atama

Koşullu ifade yalnızca yazdırmak için değil, bir değişkene değer atamak için de kullanılabilir.


````python
sayi = 7
sonuc = "çift" if sayi % 2 == 0 else "tek"
print(sonuc)
````

## 13. `in` Operatörü

`in`, bir elemanın liste, string, tuple, set veya dictionary içinde bulunup bulunmadığını kontrol eder.


````python
my_list = [1, 2, 3, 4, 5]
if 2 in my_list:
    print("3 sayısı listede var.")
````

## 14. `not` Operatörü

`not`, mantıksal sonucu tersine çevirir. `a > b` False ise `not a > b` True olur.


````python
a = 33
b = 50
if not a > b:
    print("a, b'den büyük değil.")
````

# `match-case` Yapısı

Python 3.10 ve sonrasında `match-case`, bir değerin birden fazla olası durumla karşılaştırılmasında kullanılabilir. Çok sayıda sabit değeri kontrol ederken uzun `if-elif` zincirlerine alternatif olabilir.


## 16. Gün Numarasına Göre Gün Adı


````python
gun = 6

match gun:
    case 1:
        print("Pazartesi")
    case 2:
        print("Salı")
    case 3:
        print("Çarşamba")
    case 4:
        print("Perşembe")
    case 5:
        print("Cuma")
    case 6:
        print("Cumartesi")
    case 7:
        print("Pazar")
    case _:
        print("Geçersiz gün")
````

Buradaki `case _`, önceki seçeneklerin hiçbirine uymayan durumları yakalar; `else` benzeri düşünülebilir.

## 17. Birden Fazla Değeri Aynı `case` İçinde Kullanma


````python
gun = 6

match gun:
    case 1 | 2 | 3 | 4 | 5:
        print("Hafta içi")
    case 6 | 7:
        print("Hafta sonu")
    case _:
        print("Geçersiz gün")
````

`|` işareti aynı sonucu verecek birden fazla eşleşmeyi tek `case` altında toplamamızı sağlar.

## 18. `match-case` İçinde Ek Koşul (Guard)


````python
month = 5
day = 4

match day:
    case 1 | 2 | 3 | 4 | 5 if month == 4:
        print("Hafta içi, Nisan ayı")
    case 1 | 2 | 3 | 4 | 5 if month == 5:
        print("Hafta içi, Mayıs ayı")
    case _:
        print("Diğer durum")
````

`case` ifadesinin sonuna eklenen `if` ek bir koşul (guard) oluşturur.

## Sıra Sizde 2 — Menü Uygulaması

`menu` adında bir dictionary oluşturun. Her yemek adı, fiyatıyla ilişkilendirilsin.

Program:
1. Kullanıcıdan yemek adı istesin.
2. Yemek menüde varsa fiyatını bulsun.
3. Fiyat **250 TL'den büyükse** `%10` indirim uygulasın.
4. Menüde yoksa kullanıcıya bilgi versin.

Önce kendiniz çözün.


````python
# Çözümünüzü buraya yazın

menu = {
    "pizza": 275,
    "makarna": 150,
    "hamburger": 280,
    "simit": 20
}

yemek_adi = input("Menüden yemek adı seçiniz: ").lower()

if yemek_adi in menu:
    fiyat = menu[yemek_adi]

    if fiyat > 250:
        indirimli_fiyat = fiyat * 0.90
        print(f"İndirimli fiyat: {indirimli_fiyat:.2f} TL")
    else:
        print(f"Fiyat: {fiyat:.2f} TL")
else:
    print("Seçtiğiniz yemek menüde bulunmuyor.")
````
