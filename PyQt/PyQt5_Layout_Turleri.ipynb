## 1. Layout nedir, neden ihtiyaç var?

Bir arayüzde buton, etiket ve giriş kutularını koordinatlarla tek tek yerleştirebiliriz. Ancak pencere büyüdüğünde/küçüldüğünde bu yaklaşım kırılgandır. **Layout**, widget'ların konum ve boyutlarını otomatik yöneten yerleşim sistemidir.

### Layout kullanmadan
`setGeometry(x, y, genişlik, yükseklik)` ile her widget'ın yerini elle belirleriz. Bu yönteme *absolute positioning* denir.

### Layout ile
Biz sadece “dikey sırala”, “yatay sırala”, “ızgaraya yerleştir” gibi yapıyı söyleriz; PyQt kalan hesabı yapar.

````python
import sys
from PyQt5.QtWidgets import QApplication, QWidget, QLabel, QPushButton

app = QApplication.instance() or QApplication(sys.argv)

pencere = QWidget()
pencere.setWindowTitle("Layoutsuz Arayüz")
pencere.resize(500, 300)

baslik = QLabel("Görsel Programlama", pencere)
baslik.setGeometry(30, 30, 200, 30)

buton = QPushButton("Kaydet", pencere)
buton.setGeometry(30, 80, 100, 40)

pencere.show()
# PyCharm'da dosya olarak çalıştırırken en sona: sys.exit(app.exec_())
````

### Gözlem
Pencereyi büyütüp küçültün. Widget'ların koordinatları değişmez. Bu nedenle büyük/küçük ekranlarda ve arayüz değişikliklerinde tek tek koordinat düzeltmek gerekir.

## 2. QVBoxLayout — Dikey Yerleşim
Widget'ları **yukarıdan aşağıya** dizer.

Mantık:
```text
[ Başlık ]
[ Ad     ]
[ Soyad  ]
[ Kaydet ]
```

Temel işlem: layout oluştur → widget ekle → pencereye layout'u ata.

````python
from PyQt5.QtWidgets import QWidget, QLabel, QLineEdit, QPushButton, QVBoxLayout

pencere = QWidget()
pencere.setWindowTitle("QVBoxLayout")

layout = QVBoxLayout()
layout.addWidget(QLabel("Öğrenci Bilgileri"))
layout.addWidget(QLineEdit(placeholderText="Adınız"))
layout.addWidget(QLineEdit(placeholderText="Soyadınız"))
layout.addWidget(QPushButton("Kaydet"))

pencere.setLayout(layout)
pencere.resize(400, 250)
pencere.show()
````

### Ne oldu?
`addWidget()` çağrılma sırası ekrandaki sırayı belirledi. `setGeometry()` kullanmadık. Pencere değişince layout kullanılabilir alanı widget'lara yeniden dağıtır.

## 4. Layoutları iç içe kullanmak
Gerçek arayüzler yalnızca yatay veya yalnızca dikey değildir. Bir layout başka bir layout'un içine `addLayout()` ile eklenebilir.

```text
Ad:      [...............]
Soyad:   [...............]

        [ Kaydet ] [ İptal ]
```

````python
from PyQt5.QtWidgets import (QWidget, QLabel, QLineEdit, QPushButton,
                             QVBoxLayout, QHBoxLayout)

pencere = QWidget()
pencere.setWindowTitle("İç İçe Layout")

ana_layout = QVBoxLayout()

ad_satiri = QHBoxLayout()
ad_satiri.addWidget(QLabel("Ad:"))
ad_satiri.addWidget(QLineEdit())

soyad_satiri = QHBoxLayout()
soyad_satiri.addWidget(QLabel("Soyad:"))
soyad_satiri.addWidget(QLineEdit())

butonlar = QHBoxLayout()
butonlar.addStretch()
butonlar.addWidget(QPushButton("Kaydet"))
butonlar.addWidget(QPushButton("İptal"))

ana_layout.addLayout(ad_satiri)
ana_layout.addLayout(soyad_satiri)
ana_layout.addLayout(butonlar)

pencere.setLayout(ana_layout)
pencere.resize(450, 180)
pencere.show()
````

### `addWidget()` ve `addLayout()` farkı
- `addWidget(widget)` → QLabel, QPushButton, QLineEdit gibi bir bileşen ekler.
- `addLayout(layout)` → başka bir yerleşim düzeni ekler.

## 5. QGridLayout — Satır ve Sütun Mantığı
Widget'ları tablo/ızgara şeklinde yerleştirir.

`addWidget(widget, satır, sütun)`

Satır ve sütun numaraları **0'dan başlar**.

````python
from PyQt5.QtWidgets import QWidget, QPushButton, QGridLayout

pencere = QWidget()
pencere.setWindowTitle("QGridLayout - Hesap Makinesi")

grid = QGridLayout()

metinler = [
    ("7", 0, 0), ("8", 0, 1), ("9", 0, 2),
    ("4", 1, 0), ("5", 1, 1), ("6", 1, 2),
    ("1", 2, 0), ("2", 2, 1), ("3", 2, 2),
    ("0", 3, 1)
]

for metin, satir, sutun in metinler:
    grid.addWidget(QPushButton(metin), satir, sutun)

pencere.setLayout(grid)
pencere.resize(300, 300)
pencere.show()
````

### Bir widget birden fazla hücre kaplayabilir
`addWidget(widget, satır, sütun, satırSayısı, sütunSayısı)`

Örneğin `grid.addWidget(ekran, 0, 0, 1, 4)` ekranı 4 sütuna yayar.

````python
from PyQt5.QtWidgets import QWidget, QLineEdit, QPushButton, QGridLayout

pencere = QWidget()
pencere.setWindowTitle("Grid Span")
grid = QGridLayout()

ek = QLineEdit()
ek.setPlaceholderText("Sonuç")
grid.addWidget(ek, 0, 0, 1, 4)

for i, text in enumerate(["7","8","9","+","4","5","6","-","1","2","3","×","0","C","=","÷"]):
    r = 1 + i // 4
    col = i % 4
    grid.addWidget(QPushButton(text), r, col)

pencere.setLayout(grid)
pencere.show()
````

## 6. QFormLayout — Formlar için özel layout
Kayıt, ayarlar ve veri giriş ekranlarında **etiket + alan** çiftlerini kolayca oluşturur. `QGridLayout` ile yapılabilecek işi daha okunabilir hale getirir.

````python
from PyQt5.QtWidgets import (QWidget, QLineEdit, QSpinBox, QComboBox,
                             QPushButton, QFormLayout)

pencere = QWidget()
pencere.setWindowTitle("QFormLayout")

form = QFormLayout()
form.addRow("Ad Soyad:", QLineEdit())
form.addRow("E-posta:", QLineEdit())

yas = QSpinBox()
yas.setRange(18, 100)
form.addRow("Yaş:", yas)

bolum = QComboBox()
bolum.addItems(["Yazılım Mühendisliği", "Bilgisayar Mühendisliği", "Endüstri Mühendisliği"])
form.addRow("Bölüm:", bolum)
form.addRow(QPushButton("Kaydet"))

pencere.setLayout(form)
pencere.resize(450, 250)
pencere.show()
````

## 7. Boşluk yönetimi: spacing, margins ve stretch
İyi bir arayüz yalnızca widget'ların doğru sırada olması değildir. Aralarındaki boşluk da önemlidir.

- `setSpacing(20)` → widget'lar arasındaki boşluk
- `setContentsMargins(sol, üst, sağ, alt)` → layout ile pencere kenarı arasındaki boşluk
- `addStretch()` → esneyebilen boş alan ekler
- `addWidget(widget, stretch)` → kullanılabilir alanın paylaşım oranını etkiler

````python
from PyQt5.QtWidgets import QWidget, QPushButton, QHBoxLayout

pencere = QWidget()
layout = QHBoxLayout()
layout.setSpacing(12)
layout.setContentsMargins(30, 20, 30, 20)

layout.addStretch()
layout.addWidget(QPushButton("İptal"))
layout.addWidget(QPushButton("Kaydet"))

pencere.setLayout(layout)
pencere.resize(500, 100)
pencere.show()
````

### Stretch'i deneyelim
Aşağıdaki kodda 1 ve 3 değerlerini değiştirip pencereyi büyütün. Hangi alan daha fazla genişliyor?

````python
from PyQt5.QtWidgets import QWidget, QLineEdit, QListWidget, QHBoxLayout

pencere = QWidget()
layout = QHBoxLayout()

sol = QListWidget()
sol.addItems(["Dosya 1", "Dosya 2", "Dosya 3"])
sag = QLineEdit("Detay alanı")

layout.addWidget(sol, 1)
layout.addWidget(sag, 3)
pencere.setLayout(layout)
pencere.resize(600, 250)
pencere.show()
````

## 8. Hizalama — Alignment
Bir widget'ın layout içindeki hizası `Qt.AlignLeft`, `Qt.AlignRight`, `Qt.AlignCenter`, `Qt.AlignTop`, `Qt.AlignBottom` gibi değerlerle kontrol edilebilir.

````python
from PyQt5.QtCore import Qt
from PyQt5.QtWidgets import QWidget, QLabel, QPushButton, QVBoxLayout

pencere = QWidget()
layout = QVBoxLayout()

baslik = QLabel("KULLANICI PANELİ")
layout.addWidget(baslik, alignment=Qt.AlignCenter)
layout.addWidget(QPushButton("Devam"), alignment=Qt.AlignRight)

pencere.setLayout(layout)
pencere.resize(400, 200)
pencere.show()
````

## 9. Gerçek Uygulama — Kullanıcı Kayıt Ekranı
Şimdi birden fazla layout türünü aynı arayüzde birleştirelim.

- Ana yapı → `QVBoxLayout`
- Veri giriş alanları → `QFormLayout`
- Butonlar → `QHBoxLayout`
- Butonları sağa itme → `addStretch()`

````python
import sys
from PyQt5.QtWidgets import (
    QApplication, QWidget, QLabel, QLineEdit, QComboBox,
    QPushButton, QVBoxLayout, QHBoxLayout, QFormLayout
)
from PyQt5.QtCore import Qt

class KayitPenceresi(QWidget):
    def __init__(self):
        super().__init__()
        self.setWindowTitle("Öğrenci Kayıt Sistemi")
        self.resize(500, 320)
        self.arayuzu_olustur()

    def arayuzu_olustur(self):
        ana = QVBoxLayout()
        ana.setContentsMargins(25, 25, 25, 25)
        ana.setSpacing(15)

        baslik = QLabel("ÖĞRENCİ KAYIT")
        ana.addWidget(baslik, alignment=Qt.AlignCenter)

        form = QFormLayout()
        self.no = QLineEdit()
        self.ad = QLineEdit()
        self.eposta = QLineEdit()
        self.sinif = QComboBox()
        self.sinif.addItems(["1", "2", "3", "4"])

        form.addRow("Öğrenci No:", self.no)
        form.addRow("Ad Soyad:", self.ad)
        form.addRow("E-posta:", self.eposta)
        form.addRow("Sınıf:", self.sinif)
        ana.addLayout(form)

        butonlar = QHBoxLayout()
        butonlar.addStretch()
        temizle = QPushButton("Temizle")
        kaydet = QPushButton("Kaydet")
        butonlar.addWidget(temizle)
        butonlar.addWidget(kaydet)
        ana.addLayout(butonlar)

        temizle.clicked.connect(self.formu_temizle)
        kaydet.clicked.connect(self.kaydet)
        self.setLayout(ana)

    def formu_temizle(self):
        self.no.clear()
        self.ad.clear()
        self.eposta.clear()
        self.sinif.setCurrentIndex(0)

    def kaydet(self):
        print("Öğrenci:", self.no.text(), self.ad.text(), self.eposta.text(), self.sinif.currentText())

# PyCharm'da bağımsız dosyada çalıştırmak için:
# app = QApplication(sys.argv)
# pencere = KayitPenceresi()
# pencere.show()
# sys.exit(app.exec_())
````
