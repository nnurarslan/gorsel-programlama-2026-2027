## Bölüm 1 — PyQt5, kurulum ve ilk pencere

### 1.1 GUI nedir? Neden PyQt?

CLI (Command-Line Interface), terminalden komut veya metinle etkileşimdir. GUI (Graphical User Interface), pencere, metin alanı ve butonlarla etkileşim sağlar.

**Konsol:** `input("Adınız: ")` → `print("Merhaba")`  
**GUI:** Ad giriş kutusu → **Selamla** butonu → ekranda mesaj

Qt, masaüstü arayüzleri geliştirmede kullanılan bir uygulama çatısıdır. **PyQt5**, Qt 5 sınıflarını Python'dan kullanmamızı sağlar.

> `PyQt5` ile `PyQt6` örneklerini karıştırmayın. Bu notebook boyunca `PyQt5` kullanılıyor.

### 1.2 Paket, modül, sınıf, nesne, metot, özellik

| Kavram | Tanım | PyQt5 örneği |
|---|---|---|
| Paket | İlişkili modüllerin toplandığı yapı | `PyQt5` |
| Modül | Python sınıf/fonksiyonlarını içeren birim | `PyQt5.QtWidgets` |
| Sınıf | Nesne oluşturmak için şablon | `QPushButton` |
| Nesne | Sınıfın somut örneği | `buton = QPushButton("Kaydet")` |
| Metot | Sınıfa/nesneye ait işlem | `buton.setText("Ekle")` |
| Özellik (attribute) | Nesnenin durumunu tutan değer | `self.sayac = 0` |
| Widget | Qt arayüzünde kullanılan görsel bileşen | `QLabel`, `QLineEdit` |

```text
PyQt5  (paket)
  ├── QtWidgets (modül)
  │     ├── QApplication (uygulama yöneticisi sınıfı)
  │     ├── QWidget (temel widget sınıfı)
  │     ├── QMainWindow (ana pencere sınıfı)
  │     ├── QLabel (etiket sınıfı)
  │     ├── QPushButton (buton sınıfı)
  │     └── QLineEdit (metin kutusu sınıfı)
  ├── QtCore (modül)
  └── QtGui (modül)
```

`QApplication` bir sınıftır ama ekranda görünen bir widget değildir. Uygulamanın olaylarını yönetir.

### 1.4 İlk pencerenin çalışma akışı

```text
Python programı başlar
    ↓
QApplication(sys.argv)  ← uygulama yöneticisi
    ↓
QWidget()               ← pencere nesnesi
    ↓
setWindowTitle / resize ← pencere ayarları
    ↓
show()                  ← görünür yap
    ↓
app.exec_()             ← olay döngüsünü başlat
```

- `sys.argv`: Komut satırı argümanları.
- `QApplication`: Olay döngüsü, kullanıcı etkileşimleri ve uygulama yaşam döngüsü.
- `QWidget`: Bir arayüz bileşeninin temel sınıfı; bağımsız pencere olarak da kullanılabilir.
- `show()`: Widget'ı görünür yapar.
- `app.exec_()`: PyQt5 olay döngüsünü başlatır. Programın kullanıcı etkileşimlerini işlemesini sağlar.
- `sys.exit(...)`: Uygulamanın çıkış kodunu işletim sistemine iletir.

### Örnek 1 — İlk boş pencere

Önce bu kodu çalıştırın, sonra `show()` satırını kaldırıp davranışı tartışın.

**PyCharm:** Kodu ayrı bir `.py` dosyasına yapıştırıp çalıştırın.

````python
import sys
from PyQt5.QtWidgets import QApplication, QWidget

app = QApplication(sys.argv)
pencere = QWidget()
pencere.setWindowTitle("İlk PyQt5 Pencerem")
pencere.resize(600, 350)
pencere.show()
sys.exit(app.exec_())
````

### 1.6 QWidget ile QMainWindow farkı

**QWidget:** Basit pencereler ve başka widget'ların kapsayıcısı için uygundur.

**QMainWindow:** Daha kapsamlı masaüstü uygulamalarının ana penceresi için tasarlanmıştır. Menü çubuğu, araç çubuğu, durum çubuğu ve merkezi içerik alanı desteği vardır.

```text
QWidget                    QMainWindow
+--------------------+     +--------------------------+
| Serbest içerik     |     | Menü çubuğu              |
|                    |     +--------------------------+
|                    |     | Araç çubuğu              |
|                    |     +--------------------------+
|                    |     | Merkezi içerik widget'ı  |
|                    |     +--------------------------+
+--------------------+     | Durum çubuğu             |
                           +--------------------------+
```

**Önemli:** `QMainWindow` üzerinde merkezi içerik için `setCentralWidget()` kullanılır. Bu derste merkezi `QWidget` üzerine alt widget'ları koordinatla yerleştireceğiz; layout yok.

### Örnek 2 — QMainWindow ile ilk ana pencere

`setCentralWidget()` ana içerik alanını atar; `statusBar()` durum çubuğuna erişir.

**PyCharm:** Kodu ayrı bir `.py` dosyasına yapıştırıp çalıştırın.

````python
import sys
from PyQt5.QtWidgets import QApplication, QMainWindow, QWidget, QLabel

app = QApplication(sys.argv)

pencere = QMainWindow()
pencere.setWindowTitle("QMainWindow Örneği")
pencere.resize(650, 400)

merkez = QWidget()
pencere.setCentralWidget(merkez)

etiket = QLabel("Merkezi içerik alanı", merkez)
etiket.setGeometry(30, 30, 260, 35)

pencere.statusBar().showMessage("Uygulama hazır")
pencere.show()
sys.exit(app.exec_())
````

### 1.7 Kalıtım ile pencere oluşturma

Nesne yönelimli programlamadaki kalıtımın gerçek kullanımını görelim:

- `class Pencere(QWidget)`: QWidget sınıfını miras alır.
- `__init__()`: Nesne oluşturulunca çalışan kurucu.
- `super().__init__()`: Üst sınıfın başlatılması.
- `self`: O anki pencere nesnesi.
- `self.setWindowTitle()`: Miras alınan metot.

`if __name__ == "__main__":` bloğu dosya doğrudan çalıştırıldığında uygulamayı başlatır.

### Örnek 3 — Kalıtımlı QWidget

````python
import sys
from PyQt5.QtWidgets import QApplication, QWidget

class BenimPencerem(QWidget):
    def __init__(self):
        super().__init__()
        self.setWindowTitle("Kalıtımla Pencere")
        self.setGeometry(150, 100, 520, 320)

if __name__ == "__main__":
    app = QApplication(sys.argv)
    pencere = BenimPencerem()
    pencere.show()
    sys.exit(app.exec_())
````

## Bölüm 2 — Temel widget'lar

### 2.1 Widget nedir?

Qt'de arayüzün görünen parçalarına **widget** denir. `QLabel` bir etiket, `QPushButton` buton, `QLineEdit` tek satırlık metin kutusudur.

**Parent (ebeveyn) widget:** `QLabel("Merhaba", pencere)` ifadesindeki `pencere`, etiketin üst widget'ıdır. Etiket pencerenin içerisinde gösterilir.

Bu bölümde widget'ların konumunu doğrudan belirliyoruz; layout kullanılmıyor.

### 2.2 QLabel — metin göstermek

| Metot | Açıklama |
|---|---|
| `setText("...")` | Görüntülenen yazıyı değiştirir |
| `text()` | Yazıyı okur |
| `clear()` | Yazıyı temizler |
| `setAlignment(...)` | Yazı hizasını belirler |
| `setWordWrap(True)` | Uzun metni satırlara böler |
| `setStyleSheet("...")` | Renk ve yazı tipi gibi görünümü ayarlar |

`QLabel` metin giriş alanı değildir; kullanıcıdan veri almak için `QLineEdit` kullanılır.

### 2.3 QPushButton — buton

| Metot / sinyal | Açıklama |
|---|---|
| `QPushButton("Kaydet", parent)` | Buton oluşturur |
| `setText()` | Buton yazısını değiştirir |
| `setEnabled(False)` | Butonu pasifleştirir |
| `setToolTip()` | Fare üzerine gelince ipucu gösterir |
| `clicked` | Butona tıklandığında yayılan sinyal |
| `clicked.connect(fonksiyon)` | Sinyali çalışacak fonksiyona bağlar |

**Sinyal-slot mantığı:**

```text
Kullanıcı → QPushButton'a tıklar
             ↓
         clicked sinyali
             ↓
       connect(self.islem)
             ↓
        islem() çalışır
```

**Yanlış:** `buton.clicked.connect(self.islem())` → metodu bağlantı anında çağırır.  
**Doğru:** `buton.clicked.connect(self.islem)` → metodu tıklama için bağlar.

### Örnek 5 — Buton tıklayınca metin değişsin

İlk etkileşimli uygulama. Sınıf, nesne özelliği ve olay metodu birlikte kullanılır.

**PyCharm:** Kodu ayrı bir `.py` dosyasına yapıştırıp çalıştırın.

````python
import sys
from PyQt5.QtWidgets import QApplication, QWidget, QLabel, QPushButton

class ButonPenceresi(QWidget):
    def __init__(self):
        super().__init__()
        self.setWindowTitle("Buton Olayı")
        self.resize(500, 280)

        self.mesaj = QLabel("Henüz tıklanmadı.", self)
        self.mesaj.setGeometry(40, 40, 380, 35)

        self.buton = QPushButton("Bana Tıkla", self)
        self.buton.setGeometry(40, 110, 160, 45)
        self.buton.clicked.connect(self.tiklandi)

    def tiklandi(self):
        self.mesaj.setText("Butona tıklandı!")

if __name__ == "__main__":
    app = QApplication(sys.argv)
    pencere = ButonPenceresi()
    pencere.show()
    sys.exit(app.exec_())
````

### 2.4 QLineEdit — metin girişi

| Metot / sinyal | Açıklama |
|---|---|
| `text()` | Girilen yazıyı `str` olarak döndürür |
| `setText("...")` | Giriş alanına yazı koyar |
| `clear()` | Alanı temizler |
| `setPlaceholderText("...")` | Kullanıcıya ipucu gösterir |
| `setMaxLength(20)` | En fazla karakter sayısını belirler |
| `setReadOnly(True)` | Düzenlemeyi engeller |
| `textChanged` | Yazı değişince sinyal üretir |
| `returnPressed` | Enter'a basılınca sinyal üretir |

**Önemli:** `text()` metin döndürür. Sayı istiyorsanız doğrulama ve `int()` dönüşümü gerekir.

### Örnek 6 — Adıyla selamla

`QLineEdit.text()` ile giriş alıp etikete yazdırma.

**PyCharm:** Kodu ayrı bir `.py` dosyasına yapıştırıp çalıştırın.

````python
import sys
from PyQt5.QtWidgets import QApplication, QWidget, QLabel, QLineEdit, QPushButton

class Selamlama(QWidget):
    def __init__(self):
        super().__init__()
        self.setWindowTitle("Selamlama")
        self.resize(520, 280)

        self.ad = QLineEdit(self)
        self.ad.setPlaceholderText("Adınızı giriniz")
        self.ad.setGeometry(30, 35, 280, 38)

        buton = QPushButton("Selamla", self)
        buton.setGeometry(330, 35, 140, 38)
        buton.clicked.connect(self.selamla)

        self.sonuc = QLabel("", self)
        self.sonuc.setGeometry(30, 100, 440, 40)

    def selamla(self):
        ad = self.ad.text().strip()
        if ad:
            self.sonuc.setText(f"Merhaba {ad}!")
        else:
            self.sonuc.setText("Lütfen adınızı giriniz.")

if __name__ == "__main__":
    app = QApplication(sys.argv)
    pencere = Selamlama()
    pencere.show()
    sys.exit(app.exec_())
````

## Bölüm 3 — Koordinatlarla yerleştirme

### 3.1 Koordinat sistemi

Qt'de pencere içindeki alt widget konumu, üst widget'ın **sol üst köşesine** göre belirlenir.

```text
(0,0) ───────────────────────────────→ x
  │
  │       (30,40)
  │       +------------------+
  │       | QPushButton      |  h=40
  │       +------------------+
  │           w=150
  ↓
  y
```

- `x`: Soldan uzaklık (piksel)
- `y`: Yukarıdan uzaklık (piksel)
- `width`: Genişlik
- `height`: Yükseklik

`move(x,y)` **yalnızca konumu**, `resize(w,h)` **yalnızca boyutu**, `setGeometry(x,y,w,h)` **ikisini birden** değiştirir.

**Önemli:** Pencerenin dış çerçevesi ile iç içerik alanı farklı olabilir. Üst düzey pencere koordinatları ile çocuk widget koordinatlarını karıştırmayın.

### Örnek 8 — move ve resize

İki butonu farklı konumlara yerleştiriyoruz. Bu örnekte hiçbir layout yoktur.

**PyCharm:** Kodu ayrı bir `.py` dosyasına yapıştırıp çalıştırın.

````python
import sys
from PyQt5.QtWidgets import QApplication, QWidget, QPushButton

app = QApplication(sys.argv)
pencere = QWidget()
pencere.setWindowTitle("move ve resize")
pencere.resize(600, 300)

kaydet = QPushButton("Kaydet", pencere)
kaydet.move(30, 40)
kaydet.resize(150, 45)

sil = QPushButton("Sil", pencere)
sil.move(210, 40)
sil.resize(150, 45)

pencere.show()
sys.exit(app.exec_())
````

### Örnek 10 — Bilgi kartı

Ad ve bölüm bilgilerini alıp etiket üzerinde gösteren, layoutsuz uygulama.

**PyCharm:** Kodu ayrı bir `.py` dosyasına yapıştırıp çalıştırın.

````python
import sys
from PyQt5.QtWidgets import QApplication, QWidget, QLabel, QLineEdit, QPushButton

class BilgiKarti(QWidget):
    def __init__(self):
        super().__init__()
        self.setWindowTitle("Kullanıcı Bilgi Kartı")
        self.resize(600, 330)

        QLabel("Ad Soyad:", self).setGeometry(30, 35, 110, 35)
        self.ad = QLineEdit(self)
        self.ad.setGeometry(150, 35, 360, 35)

        QLabel("Bölüm:", self).setGeometry(30, 90, 110, 35)
        self.bolum = QLineEdit(self)
        self.bolum.setGeometry(150, 90, 360, 35)

        goster = QPushButton("Bilgileri Göster", self)
        goster.setGeometry(150, 150, 180, 45)
        goster.clicked.connect(self.goster)

        self.sonuc = QLabel("Bilgiler burada görünecek.", self)
        self.sonuc.setGeometry(30, 225, 520, 45)

    def goster(self):
        ad = self.ad.text().strip()
        bolum = self.bolum.text().strip()
        if not ad or not bolum:
            self.sonuc.setText("Lütfen iki alanı da doldurun.")
            return
        self.sonuc.setText(f"{ad} — {bolum}")

if __name__ == "__main__":
    app = QApplication(sys.argv)
    pencere = BilgiKarti()
    pencere.show()
    sys.exit(app.exec_())
````
